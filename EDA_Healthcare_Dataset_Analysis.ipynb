```python
df["Length of stay"] = (
    pd.to_datetime(df["Discharge Date"]) - 
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission","Discharge Date","Length of stay"]].head() 
```

```python
df["Admission Year"] = pd.to_datetime(
    df["Date of Admission"]
).dt.year

df[["Date of Admission", "Admission Year"]].head() 
```

```python
df["Admission Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.month

df[["Date of Admission", "Admission Month"]].head() 
```

```python
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
monthly_admissions = (
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)

monthly_admissions 
```

```python
plt.figure(figsize=(9, 5))

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Name",
    marker="o"

)

plt.title("Monthly Number of Admission")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()           
```
